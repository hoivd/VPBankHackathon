```python
from tqdm import tqdm
import time
import logging

def batch_embed(texts, batch_size=128, max_retries=3, retry_delay=1):
    all_embeddings = []
    for i in tqdm(range(0, len(texts), batch_size), desc="Embedding batches"):
        k = i + batch_size
        print(f"Embedding batch from {i} to {k}")
        batch = texts[i:k]

        retry_count = 0
        while retry_count < max_retries:
            try:
                embeddings = embedder.embed(batch)  # gọi theo batch
                all_embeddings.extend(embeddings)
                break  # thành công -> thoát retry loop
            except Exception as e:
                retry_count += 1
                logging.warning(f"Lỗi khi embed batch {i}-{k}: {e}. Thử lại lần {retry_count}/{max_retries}...")
                time.sleep(retry_delay * retry_count)  # tăng dần thời gian chờ

                if retry_count == max_retries:
                    logging.error(f"Thất bại sau {max_retries} lần retry cho batch {i}-{k}. Bỏ qua batch này.")
                    all_embeddings.extend([[0.0]*1024 for _ in batch])  # hoặc append vector zero placeholder nếu cần

    return all_embeddings
```

```python
def batch_faiss_search_from_embeddings(embeddings: np.ndarray, faiss_searcher, top_k: int = 10):
    """
    Tìm top_k kết quả từ FAISS cho batch embedding đã chuẩn bị sẵn (đã normalize nếu cần).

    Parameters:
        embeddings (np.ndarray): (n_query, dim), đã normalize nếu dùng inner product.
        faiss_searcher: Đối tượng FaissSearcher có phương thức search_top_k.
        top_k (int): Số kết quả cần trả về cho mỗi truy vấn.

    Returns:
        list[list[dict]]: Mỗi phần tử là danh sách kết quả top_k cho 1 embedding.
    """
    if not isinstance(embeddings, np.ndarray):
        embeddings = np.array(embeddings, dtype='float32')

    if embeddings.ndim == 1:
        embeddings = embeddings.reshape(1, -1)

    embeddings = embeddings.astype('float32')  # chỉ cần đảm bảo đúng dtype

    all_results = []
    for i in tqdm(range(len(embeddings)), desc="FAISS batch search"):
        result = faiss_searcher.search_top_k(embeddings[i].reshape(1, -1), top_k=top_k)
        all_results.append(result)

    return all_results
```

```python
def evaluate_faiss_search(
    faiss_results: list[list[dict]],
    ground_truth_ids: list[int],
    queries: list[str] = None,
    k: int = 10,
    verbose: bool = True
):
    """
    Đánh giá chi tiết FAISS search với ground-truth, hiển thị đúng/sai và thống kê.

    Parameters:
        faiss_results (list[list[dict]]): Kết quả từ FAISS search.
        ground_truth_ids (list[int]): ID đúng của từng truy vấn.
        queries (list[str]): Danh sách câu truy vấn gốc.
        k (int): Top-k để đánh giá.
        verbose (bool): Nếu True, in kết quả đúng/sai chi tiết.

    Returns:
        dict: Gồm accuracy, correct_indices, incorrect_indices.
    """
    correct_indices = []
    incorrect_indices = []

    for i, (result, gt_id) in enumerate(zip(faiss_results, ground_truth_ids)):
        top_k_ids = [hit['id'] for hit in result[:k]]
        if gt_id in top_k_ids:
            correct_indices.append(i)
        else:
            incorrect_indices.append(i)

    accuracy = len(correct_indices) / len(ground_truth_ids) if ground_truth_ids else 0.0

    print(f"\n📊 Đánh giá Top-{k} Accuracy: {accuracy:.2%}")
    # if verbose:
    #     print(f"\n📊 Đánh giá Top-{k} Accuracy: {accuracy:.2%}")
    #     print(f"✅ Truy vấn đúng: {len(correct_indices)} / {len(ground_truth_ids)}")
    #     print(f"❌ Truy vấn sai: {len(incorrect_indices)} / {len(ground_truth_ids)}")

    #     if correct_indices:
    #         print("\n✅ Các truy vấn đúng:")
    #         for i in correct_indices:
    #             print(f"  [✓] Query {i+1}: \"{queries[i]}\" → Ground-truth ID: {ground_truth_ids[i]}")
        
    #     if incorrect_indices:
    #         print("\n❌ Các truy vấn sai:")
    #         for i in incorrect_indices:
    #             top_ids = [hit['id'] for hit in faiss_results[i][:k]]
    #             print(f"  [✗] Query {i+1}: \"{queries[i]}\" → Ground-truth: {ground_truth_ids[i]} NOT IN {top_ids}")

    return {
        'accuracy': accuracy
        # 'correct': correct_indices,
        # 'incorrect': incorrect_indices
    }
```

```python
import random

def random_drop_any_field(data: dict) -> dict:
    """
    Random xóa 1 trường bất kỳ trong dict, kể cả 'last_full_name'.

    Parameters:
        data (dict): Dictionary đầu vào.

    Returns:
        dict: Bản sao của dict đã bị xóa ngẫu nhiên 1 trường.
    """
    if not data:
        return data.copy()

    keys = list(data.keys())
    key_to_drop = random.choice(keys)

    new_data = data.copy()
    del new_data[key_to_drop]

    return new_data

def apply_random_drop_to_list(data_list: list[dict]) -> list[dict]:
    """
    Áp dụng random xóa 1 trường cho mỗi dict trong danh sách.

    Parameters:
        data_list (list[dict]): Danh sách các dict.

    Returns:
        list[dict]: Danh sách mới sau khi mỗi dict bị xóa 1 trường ngẫu nhiên.
    """
    return [random_drop_any_field(item) for item in data_list]
```

```python
def evaluate_multiple_ks(faiss_results, ground_truth_ids, queries, ks=[1, 10, 20]):
    """
    Chạy đánh giá FAISS search với nhiều giá trị k (top-k).
    
    Parameters:
        faiss_results (list[list[dict]]): Kết quả từ FAISS search.
        ground_truth_ids (list[int]): Danh sách ID đúng.
        queries (list[str]): Câu truy vấn gốc.
        ks (list[int]): Danh sách các giá trị k muốn đánh giá.
    """
    print("📌 Đánh giá FAISS với nhiều mức Top-k:\n")
    results_summary = {}
    for k in ks:
        print(f"\n==================== 🔎 Top-{k} ====================")
        result = evaluate_faiss_search(
            faiss_results=faiss_results,
            ground_truth_ids=ground_truth_ids,
            queries=queries,
            k=k,
            verbose=True
        )
        results_summary[k] = result['accuracy']
    
    # Tổng hợp nhanh:
    print("\n📈 Tổng kết accuracy theo k:")
    for k in ks:
        print(f"  Top-{k} Accuracy: {results_summary[k]:.2%}")
```
