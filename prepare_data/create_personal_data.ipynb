# Random viet tat

```python
import unicodedata
import random

def remove_accents(text):
    return ''.join(c for c in unicodedata.normalize('NFD', text)
                   if unicodedata.category(c) != 'Mn')

def random_abbreviation(full_name):
    parts = full_name.strip().split()
    if not parts or len(parts) < 2:
        return full_name

    rules = []

    # Rule 1: Chữ cái đầu mỗi phần
    rules.append(" ".join([p[0] for p in parts]))

    # Rule 2: 2 chữ cái đầu mỗi phần
    rules.append(" ".join([p[:2] for p in parts]))

    # Rule 3: Giữ tên chính (cuối), rút gọn phần còn lại
    rules.append(" ".join([p[0] for p in parts[:-1]] + [parts[-1]]))
    rules.append(" ".join([p[:2] for p in parts[:-1]] + [parts[-1]]))

    # Rule 6: Không dấu, viết liền
    rules.append(remove_accents("".join(parts)))

    # Rule 7: Giữ thứ tự, mỗi phần có thể là 1 chữ cái, 2 chữ cái, hoặc giữ nguyên
    combo = []
    for p in parts:
        choice = random.choice([p[0], p[:2], p])
        combo.append(choice)
    rules.append(" ".join(combo))

    # Rule 8: Ghép liền, không dấu cách
    rules.append("".join([p[0] for p in parts]))  # ví dụ: NVA
    rules.append("".join([p[:2] for p in parts]))  # ví dụ: NgVaAn

    # Rule 9: Chữ đầu hoa, còn lại thường
    rules.append(" ".join([p.capitalize() for p in parts]))

    # Random chọn một trong các quy tắc
    return random.choice(rules)
```

```python
def random_last_name(row, full_prob=0.7):
    if random.random() < full_prob:
        return row['full_name']
    else:
        return row['abbreviation']

df_all['last_full_name'] = df_all.apply(random_last_name, axis=1)
```
