````python
from sentence_transformers import SentenceTransformer
import faiss
import numpy as np

# 1. Load pre-trained model để tạo embedding
model = SentenceTransformer("sentence-transformers/paraphrase-multilingual-MiniLM-L12-v2")

# 2. Tập văn bản mẫu

# 3. Encode văn bản thành embedding vectors
embeddings = model.encode(contents)

# 4. Convert sang dạng float32 cho FAISS
embeddings = np.array(embeddings).astype("float32")

# 5. Tạo FAISS index
dimension = embeddings.shape[1]
index = faiss.IndexFlatL2(dimension)  # dùng L2 (Euclidean); có thể dùng IndexFlatIP cho cosine similarity
index.add(embeddings)  # thêm dữ liệu vào index

# 6. Truy vấn văn bản mới
query = new_text
query_vector = model.encode([query]).astype("float32")

# 7. Tìm top 3 văn bản gần nhất
k = 5
distances_content, indices_content = index.search(query_vector, k)
````

````python
import json
import re

# Giả sử đây là nội dung phản hồi từ Gemini (chứa 2 khối JSON)
raw_text = resonse_28_risk.text

# Tìm tất cả các block JSON được bao trong ```json ... ``` markdown
json_blocks = re.findall(r"```json\s*\n(.*?)```", raw_text, re.DOTALL)

# Đảm bảo có đúng 2 khối JSON
if len(json_blocks) >= 2:
    personal_info_json = json.loads(json_blocks[0])
    risk_info_json = json.loads(json_blocks[1])
else:
    raise ValueError("Không tìm thấy đủ hai khối JSON trong đoạn văn bản.")

# ✅ In kết quả kiểm tra
print("🔹 Personal Info JSON:")
print(json.dumps(personal_info_json, indent=2, ensure_ascii=False))

print("\n🔹 Risk Info JSON:")
print(json.dumps(risk_info_json, indent=2, ensure_ascii=False))
````
